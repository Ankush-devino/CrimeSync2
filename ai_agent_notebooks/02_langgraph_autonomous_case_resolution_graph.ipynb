```python
# bansal
import json
from typing import Dict, Any, List, TypedDict

print("[+] Compiling CrimeSync LangGraph State Machine (LangGraph-Runtime-v0.2)")
```

### 1. State Definition and Node Handlers

```python
# bansal
class InvestigationState(TypedDict):
    case_id: str
    leads: List[str]
    evidence_hashes: List[str]
    suspect_risk_score: float
    warrant_requested: bool
    warrant_approved: bool
    investigation_complete: bool

def analyze_leads_node(state: InvestigationState) -> InvestigationState:
    print(f"[*] [Node: Analyze Leads] Evaluating {len(state['leads'])} evidence leads for {state['case_id']}")
    state['suspect_risk_score'] = 92.5
    state['evidence_hashes'].append("e3b0c44298fc1c149afbf4c8996fb92427ae41e4649b934ca495991b7852b855")
    return state

def assess_warrant_threshold_node(state: InvestigationState) -> InvestigationState:
    print(f"[*] [Node: Assess Warrant] Suspect Risk: {state['suspect_risk_score']} / 100")
    if state['suspect_risk_score'] > 85.0:
        state['warrant_requested'] = True
        print("    -> Triggering Judicial Warrant Request under Section 94 BNSS (Search & Seizure)")
    return state

def judicial_approval_node(state: InvestigationState) -> InvestigationState:
    print("⚖️ [Node: Human-in-the-Loop Judicial Approval] Magistrate review of Section 65B blockchain proof...")
    state['warrant_approved'] = True
    state['investigation_complete'] = True
    return state

class LangGraphExecutor:
    def run(self, initial_state: InvestigationState) -> InvestigationState:
        s = analyze_leads_node(initial_state)
        s = assess_warrant_threshold_node(s)
        if s['warrant_requested']:
            s = judicial_approval_node(s)
        return s

graph = LangGraphExecutor()
init_state: InvestigationState = {
    "case_id": "CASE-2026-011",
    "leads": ["WhatsApp extortion audio", "ICICI Mule Account Statement", "Honeypot TLS Fingerprint"],
    "evidence_hashes": [],
    "suspect_risk_score": 0.0,
    "warrant_requested": False,
    "warrant_approved": False,
    "investigation_complete": False
}

final_state = graph.run(init_state)
print("\n[+] LangGraph Execution Completed:")
print(json.dumps(final_state, indent=2))
```
